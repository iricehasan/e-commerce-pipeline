# Select Catalog

```python
%sql
CREATE CATALOG IF NOT EXISTS ecommerce_project;
USE CATALOG ecommerce_project;
```

# Create Lakehouse Schemas

```python
%sql
CREATE SCHEMA IF NOT EXISTS bronze
COMMENT 'Bronze layer: raw data landed as-is, no cleaning';

CREATE SCHEMA IF NOT EXISTS silver
COMMENT 'Silver layer: deduplicated, validated, one row per real event';

CREATE SCHEMA IF NOT EXISTS gold
COMMENT 'Gold layer: business-facing aggregates';
```

# Create Volume

```python
%sql
CREATE VOLUME IF NOT EXISTS bronze.raw_files
COMMENT 'Landing zone for the data generator''s Parquet output, read by bronze.py';
```
